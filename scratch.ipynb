```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from config import API_MODEL
import numpy as np
```

```python
from config import EMBEDDING_MODEL


load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    raise ValueError("OPENAI_API_KEY environment variable is not set.")
else:
    print("OPENAI_API_KEY environment variable is set.")

openAI = OpenAI()
model = API_MODEL
embedding_model = EMBEDDING_MODEL
```

```python
response =openAI.embeddings.create(input=[ "My kid is cute", "Hello, my dog is cute"], model=EMBEDDING_MODEL)
data = response.data
vectors = []

for d in data:
    vectors.append(d.embedding)
    # print(d.embedding)
    # print(len(d.embedding))
    # print(d.index)

print(np.array(vectors, dtype=np.float32))
```
